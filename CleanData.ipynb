```python
# 1 — Setup: mount Drive, Identify the path and filename of the CSV file within the zip archive.

from google.colab import drive
import pandas as pd
import numpy as np
import zipfile

pd.set_option("display.max_columns", None)
SEED = 42
np.random.seed(SEED)
drive.mount('/content/drive')
zip_path = '/content/drive/MyDrive/ADY201m/US_Accidents_March23.zip'
with zipfile.ZipFile(zip_path, 'r') as z:
    csv_filename = [name for name in z.namelist() if name.endswith('.csv')][0]
print(f"File CSV bên trong zip: {csv_filename}")
```

```python
# 2 — Read data from the ZIP file using chunking, reading only the columns required for the data cleaning stage.

cols_needed = [
    # Supplementary
    "ID", "Source", "Severity", "Description",
    # Weather
    "Weather_Timestamp", "Temperature(F)", "Wind_Chill(F)", "Humidity(%)",
    "Pressure(in)", "Visibility(mi)", "Wind_Direction", "Wind_Speed(mph)",
    "Precipitation(in)", "Weather_Condition",
    # Time
    "Start_Time", "End_Time", "Sunrise_Sunset", "Civil_Twilight",
    "Nautical_Twilight", "Astronomical_Twilight",
    # Geographical Location
    "Start_Lat", "Start_Lng", "End_Lat", "End_Lng", "Distance(mi)",
    "Street", "City", "County", "State", "Zipcode", "Country", "Timezone", "Airport_Code",
    # Road Infrastructure
    "Amenity", "Bump", "Crossing", "Give_Way", "Junction", "No_Exit",
    "Railway", "Roundabout", "Station", "Stop", "Traffic_Calming",
    "Traffic_Signal", "Turning_Loop",
]
CHUNK_SIZE = 500_000
chunks_list = []
total_rows_read = 0
print("Starting to read data using chunking...")
with zipfile.ZipFile(zip_path, 'r') as z:
    with z.open(csv_filename) as f:
        for i, chunk in enumerate(pd.read_csv(f, usecols=cols_needed, chunksize=CHUNK_SIZE)):
            chunks_list.append(chunk)
            total_rows_read += len(chunk)
            print(f"  Chunk {i+1}: read {total_rows_read:,} rows")
df = pd.concat(chunks_list, ignore_index=True)
del chunks_list
print(f"\nInitial data size: {df.shape}")
df.head()
```

```python
# 3 — Remove columns not used for analysis/modeling.
cols_to_drop = [
    "ID", "Distance(mi)", "Description",
    "Wind_Chill(F)", "Precipitation(in)",
    "Country",
    "End_Lat", "End_Lng",
    "Roundabout", "Turning_Loop",
    "Traffic_Calming", "Bump",
]
df = df.drop(columns=cols_to_drop)
print(f"Dimensions after column removal: {df.shape}")
print(f"The remaining columns:\n{list(df.columns)}")

print("\n The first two rows of the data after deleting the column. ")
print(df.head(2))
```

```python
# 4 — Cast Start_Time and End_Time to datetime + forward fill

df["Start_Time"] = pd.to_datetime(df["Start_Time"], errors="coerce")
df["End_Time"] = pd.to_datetime(df["End_Time"], errors="coerce")
print(f"Số dòng Start_Time resulted in an error/missing value (NaT) before filling: {df['Start_Time'].isna().sum()}")
print(f"Số dòng End_Time resulted in an error/missing value (NaT) before filling: {df['End_Time'].isna().sum()}")
df["Start_Time"] = df["Start_Time"].ffill()
df["End_Time"] = df["End_Time"].ffill()
print(f"Number of missing rows after forward fill: "
      f"Start_Time={df['Start_Time'].isna().sum()}, End_Time={df['End_Time'].isna().sum()}")
```

```python
# 5 — 4 Twilight variants: null type, Day/Night converted to 0/1.

twilight_cols = ["Sunrise_Sunset", "Civil_Twilight", "Nautical_Twilight", "Astronomical_Twilight"]
before = len(df)
df = df.dropna(subset=twilight_cols)
after = len(df)
print(f"{before - after:,} rows were excluded due to missing twilight data. ({(before-after)/before*100:.2f}%)")
for col in twilight_cols:
    df[col] = df[col].map({"Day": 1, "Night": 0})
    if df[col].isna().sum() > 0:
        mode_val = df[col].mode()[0]
        df[col] = df[col].fillna(mode_val)
        print(f"  {col}: filled {df[col].isna().sum()} missing values ​​with the mode = {mode_val}")
print(df[twilight_cols].head())
```

```python
# 6 — Weather variable group: null imputation (<5%), quartile imputation for Wind_Speed.

weather_drop_null_cols = [
    "Weather_Timestamp", "Temperature(F)", "Humidity(%)",
    "Pressure(in)", "Visibility(mi)", "Wind_Direction", "Weather_Condition"
]
before = len(df)
df = df.dropna(subset=weather_drop_null_cols)
after = len(df)
print(f"Đã loại {before - after:,} dòng do thiếu dữ liệu thời tiết cơ bản ({(before-after)/before*100:.2f}%)")
missing_before = df["Wind_Speed(mph)"].isna().sum()
pct_missing = missing_before / len(df) * 100
print(f"Wind_Speed(mph) thiếu {missing_before:,} dòng ({pct_missing:.2f}%) trước khi điền")
median_wind_speed = df["Wind_Speed(mph)"].median()
df["Wind_Speed(mph)"] = df["Wind_Speed(mph)"].fillna(median_wind_speed)
print(f"Đã điền {missing_before:,} giá trị thiếu bằng median = {median_wind_speed:.2f} mph")
```

```python
# 7 — Geographic location variable group: enter "Unknown" for text variables.

location_fill_unknown_cols = ["Street", "City", "Zipcode", "Timezone", "Airport_Code"]
for col in location_fill_unknown_cols:
    missing_count = df[col].isna().sum()
    df[col] = df[col].fillna("Unknown")
    print(f"{col}: filled {missing_count:,} missing value by 'Unknown'")
print("\n Check the remaining position columns.")
print(df[["Start_Lat", "Start_Lng", "County", "State"]].isna().sum())
```

```python
# 8 — Road infrastructure variable group (boolean)
infra_cols = ["Amenity", "Crossing", "Give_Way", "Junction", "No_Exit",
              "Railway", "Station", "Stop", "Traffic_Signal"]

print(" Check for missing values ​​in the road infrastructure group. ")
print(df[infra_cols].isna().sum())

for col in infra_cols:
    if df[col].isna().sum() > 0:
        mode_val = df[col].mode()[0]
        df[col] = df[col].fillna(mode_val)
        print(f"  {col}: filled by mode = {mode_val}")
```

```python
# 9 — Final check

print("=" * 70)
print("Check after cleaning")
print("=" * 70)
print(f"\nShape of final data: {df.shape}")
missing_summary = df.isna().sum()
missing_summary = missing_summary[missing_summary > 0]
if len(missing_summary) == 0:
    print("✅ Not found missing value in dataset!")
else:
    print(missing_summary)
print(f"\ndata type of each column:")
print(df.dtypes)
df.head()
```
